#Different Opperation


```python
a=tf.constant([1,2]) # 1 Rows , 2 Columns
b=tf.constant([3,4]) # 1 Rows , 2 Columns
```

```python
# Sum
c=tf.add(a,b)
c
```

```python
# Subtract
d=tf.subtract(a,b)
d
```

```python
# Multiplication
a_1 = tf.constant([[1,2], [3,4]])
b_1 = tf.constant([[5,6], [7,8]])

print(f'a_1 matrix is:\n{a_1}\n \n b_1 matrix is :\n{b_1}')
```

```python
ew = tf.multiply(a_1, b_1)
ew
```

```python
nm= tf.matmul(a_1,b_1)
nm
```

```python
# Division
a=tf.constant([1,2])
b=tf.constant([3,0])

d=tf.divide(a,b)
d
```

#Operation On Element

```python
a=tf.constant([[1,2,3],
              [4,5,6]])
a
```

```python
#Square
b=tf.square(a)
b
```

```python
#sqrt
c=tf.sqrt(tf.cast(a,dtype=float))
c
```

#Min, Max, Mean, STD

```python
a=tf.constant([[1,2,3],[4,5,6]])
a
```

```python
#MIN
min = tf.reduce_min(a,axis=1)
min
```

```python
#MAX
max = tf.reduce_max(a,axis=1)
max
```

```python
#AVERAGE
avg = tf.reduce_mean(a,axis=1)
avg
```

```python
#standard deviation
std = tf.math.reduce_std(tf.cast(a,float))
std
```

```python
#SUM
sum = tf.reduce_sum(a, axis=1)
sum
```
