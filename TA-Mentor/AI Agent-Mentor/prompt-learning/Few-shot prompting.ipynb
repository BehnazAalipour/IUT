```python
import os
from dotenv import load_dotenv
from langchain_openai import OpenAI

from langchain.prompts import PromptTemplate
from langchain.prompts import ChatPromptTemplate
from langchain.schema.output_parser import StrOutputParser
```

```python
load_dotenv()
```

```python
api_key= os.getenv("OPENAI_API_KEY")
llm=OpenAI(model="gpt-4o-mini")

from langchain.prompts import FewShotChatMessagePromptTemplate

examples = [
    {"q": "What is overfitting?", "a": "Overfitting is when a model memorizes noise."},
    {"q": "What is underfitting?", "a": "Underfitting is when a model is too simple."},
]

example_prompt = ChatPromptTemplate.from_messages([
    ("human", "{q}"),
    ("ai",    "{a}")
])

fewshot = FewShotChatMessagePromptTemplate(
    examples=examples,
    example_prompt=example_prompt
)

main = ChatPromptTemplate.from_messages([
    ("system", "Answer briefly, like the examples."),
    fewshot,
    ("human", "{question}")
])

print((main | llm | StrOutputParser()).invoke({"question": "What is cross-validation?"}))
```
